#### Look for null or missing data and take the best option at this case

```python
engine = get_engine()

df = pd.read_sql("SELECT * FROM products", con=engine)
print(df.isnull().sum())
```

```python
df_items = pd.read_sql("SELECT * FROM order_items", con=engine)

null_products = df.loc[df["product_category_name"].isnull(),
                       'product_id'
                       ].tolist()

df_related = df_items[
    df_items['product_id'].isin(null_products)
]

print(df_related['order_id'].count())
```
